## DATA CLEANING SCRIPT

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv("C:/Users/ADMIN/Desktop/PROJECTS/Luxury Housing Sales Analysis Project/Luxury_Housing_Bangalore.csv")
```

```python
df.columns = df.columns.str.strip().str.replace(" ", "_").str.replace("-", "_")
```

```python
df.columns
```

### HANDLING MISSING VALUES

```python
df.info()
```

```python
df.isnull().sum()
```

#### FILLING MISSING FIELDS

```python
num_cols = ['Unit_Size_Sqft','Ticket_Price_Cr','Amenity_Score']
for col in num_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(r'[^0-9\.]', '', regex=True)  
        .replace('', None))
    
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
print(df[num_cols].isnull().sum())
```

```python
df[num_cols]
```

```python
df
```

```python
text_cols = df.select_dtypes(include='object').columns
df[text_cols].isnull().sum()
```

```python
text_cols = ['Buyer_Comments']
for col in text_cols:
    if col in df.columns:
        df[col].fillna('Unknown', inplace=True)
```

```python
text_cols = df.select_dtypes(include='object').columns
df[text_cols].isnull().sum()
```

#### NORMALIZATION

```python
df['Developer_Name'].unique()
```

```python
df['Developer_Name'] = df['Developer_Name'].astype(str).str.upper()
df['Micro_Market'] = df['Micro_Market'].astype(str).str.upper()
df['Configuration'] = df['Configuration'].astype(str).str.upper()
```

#### ADD COLUMNS

```python
df['Transaction_Type'].unique()
```

```python
df['Price_per_Sqft'] = df['Ticket_Price_Cr'] / df['Unit_Size_Sqft']

df['Purchase_Quarter'] = pd.to_datetime(df['Purchase_Quarter'], errors = 'coerce')

df['Quarter_Number'] = df['Purchase_Quarter'].dt.quarter

df['Year'] = df['Purchase_Quarter'].dt.year

df['Booking_Flag'] = df['Transaction_Type'].apply(
    lambda x: 1 if str(x).strip() == 'Primary' else 0)
```
